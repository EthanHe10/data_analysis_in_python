## 2. Series reindexing

given a Series s1, s1.reindex() function takes an array of index as the argument, then returns a new Series.

```python
s1 = pd.Series([4.5, 7.2, -5.3, 3.6], index = ["d", "b", "a", "c"])

s1
```

```python
# s1.reindex() will create a new Series

s2 = s1.reindex(["a", "b", "c", "d"])

s2
```

```python
s3 = s1.reindex(["a", "b", "c", "d", "e"], fill_value = 999)

s3
```

# 4. Apply a function to a Series

Given a Series s, s.map(f) returns a new Series that executed function f element-wise of s

```python
def absolute(x):
    if x >= 0:
        return x
    else:
        return -x
```

```python
absolute(-99)
```

```python
s1 = pd.Series(np.random.randn(5), index = np.arange(5)+1)

s1
```

```python
s2 = s1.map(absolute)
s2
```

```python
s1 # s1 doesn't change
```

## Data Frame Reindexing

```python
import numpy as np
import pandas as pd
```

given a data frame df1

1. df1.reindex() takes a list as an argument
2. it returns a NEW data frame, and the df1 doesn't change.
3. you can reindex index, or you can reindex the columns.

```python
df1 = pd.DataFrame(np.arange(9).reshape(3,3),
                   index = ["a", "c", "d"],
                   columns = ["Ohio", "Texas", "California"])

df1
```

```python
df2 = df1.reindex(["a", "b", "c", "d"])

df2
```

```python
df1
```

```python
# reindex the columns

new_columns = ["Ohio", "Utah", "California"]

df3 = df1.reindex(columns = new_columns)

df3
```

## 12. Drop Items or Columns from a Data Frame

given a data frame df1
1. df1.drop() takes a list as an argument
2. axis = 1 means drop columns
3. the method returns a new edata frame, the df1 doesn't change

```python
df1 = pd.DataFrame(np.arange(16).reshape(4,4),
                   index = ['Ohio', 'Colorado', 'Utah', 'New York'],
                   columns = ['one', 'two', 'three', 'four'])

df1
```

```python
df2 = df1.drop(["Ohio", "Utah"])

df2
```

```python
df3 = df1.drop(["one", "three"], axis = 1)

df3
```
