## 1. Basic Data Structure: Series

1. A series object has 2 columns: index, and corresponding values.
2. Series is similar to python dictionary.
3. We can use python dictionary to create a Series object.

```python
import numpy as np

import pandas as pd
```

```python
# Create a Series

s1 = pd.Series([2,3,4])

s1
```

```python
s1.index # it returns the series s1's index
```

```python
s1.values # it returns the series s1's values as an array
```

```python
# create a series with specific index

s2 = pd.Series([2,3,4], index=["a","b","c"])

s2
```

```python
s2["c"] # it returns the value whose index is "c"
```

```python
s2[["a","b"]] # it returns part of the Series s2 whose index are taken in
```

```python
# check if a specific index is in a given series

"c" in s2
```

```python
"f" in s2
```

```python
# Create a Series from a python dictionary

dict1 = {'Ohio':35000, 'Texas':71000, 'Oregon': 16000, 'Utah': 5000}

dict1
```

```python
s1 = pd.Series(dict1)

s1
```

```python
# we can build a new Series based on an existing one

states = ['California', 'Ohio', 'Oregon', 'Texas']

s2 = pd.Series(dict1, index = states)

s2
```

```python
s2["California"] = 22223

s2
```

```python
states = ['New York', 'Ohio', 'Oregon', 'Texas']

s3 = pd.Series(dict1, index = states)

s3
```

```python
pd.notnull(s3)
```

```python
pd.isnull(s3)
```

```python
s1 + s2
```

```python
s1 + s2 + s3
```

```python
s1 * 2
```

```python
np.exp(s1/10000) # element-wise exponential function value
```

## 2. Series reindexing

given a Series s1, s1.reindex() function takes an array of index as the argument, then returns a new Series.

```python
s1 = pd.Series([4.5, 7.2, -5.3, 3.6], index = ["d", "b", "a", "c"])

s1
```

```python
# s1.reindex() will create a new Series

s2 = s1.reindex(["a", "b", "c", "d"])

s2
```

```python
s3 = s1.reindex(["a", "b", "c", "d", "e"], fill_value = 999)

s3
```

## 3. Drop entries from a Series

given a Series s, use a s.drop() funcion

s.drop() takes an array of index you want to drop as the argument

it returns a new series, the original series doesn't change

```python
s1 = pd.Series(np.arange(5.0), index=['a','b','c','d','e'])

s1
```

```python
s2 = s1.drop(["b", "d"])

s2
```

```python
s1
```

# 4. Apply a function to a Series

Given a Series s, s.map(f) returns a new Series that executed function f element-wise of s

```python
def absolute(x):
    if x >= 0:
        return x
    else:
        return -x
```

```python
absolute(-99)
```

```python
s1 = pd.Series(np.random.randn(5), index = np.arange(5)+1)

s1
```

```python
s2 = s1.map(absolute)
s2
```

```python
s1 # s1 doesn't change
```

## 5. Series Sorting adn Ranking

Given a Series s

s.sort_index(): sorted by index, return a new Series, the original s doesn't change

s.sort_value(): sorted by values, return a new Series, the original s doesn't change

```python
s = pd.Series(np.random.randn(5), index = np.random.permutation(np.arange(5)+1))

s
```

```python
s1 = s.sort_index()

s1
```

```python
s2 = s.sort_values()

s2
```
