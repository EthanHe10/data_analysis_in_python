```python
# Create an array with zeros

arr5 = np.zeros(8) # It will return a one-dimensioanl array with 8 zeros

arr5
```

```python
# Create a 2-dimensioanl array of shape (m, n) with zeros

arr6 = np.zeros((3,4)) # it returns a array of shape (3,4) with zeros

arr6
```

```python
# np.arange () it's similar to range() in list

np.arange(10) # it returns an array of elements from 0 to 9, 10 elements in total
```

```python
# nparange() generates an arithmetic sequence by indicating the start, stop and step.

np.arange(-3, 5, 0.5)

# It returns an arithmetic sequence starting with -3, and common difference d = 0.5, it stops before 5 (5 is not included)
```

```python
# np.arange() the default tep is 1 if not specified.

np.arange(-3,5)

# It will return an arithmetic sequence from -3 to 4 by common different d = 1 (default)
```

## 2. Operations between Arrays

```python
import numpy as np
```

```python
arr1 = np.arange(1, 9).reshape((2,4))

arr1
```

```python
np.arange(1, 9)
```

```python
np.arange(1, 9).reshape((2,4))
```

```python
arr2 = np.arange(8).reshape((2,4))

arr2
```

```python
arr1 + arr2
```

```python
# array multiplication: element-wise multiplication

arr1 * arr2
```

```python
1/arr1
```
