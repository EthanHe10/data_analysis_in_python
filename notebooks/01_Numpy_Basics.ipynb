```python
# nparange() generates an arithmetic sequence by indicating the start, stop and step.

np.arange(-3, 5, 0.5)

# It returns an arithmetic sequence starting with -3, and common difference d = 0.5, it stops before 5 (5 is not included)
```

```python
# np.arange() the default tep is 1 if not specified.

np.arange(-3,5)

# It will return an arithmetic sequence from -3 to 4 by common different d = 1 (default)
```
